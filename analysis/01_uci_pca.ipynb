# 01 UCI gas sensor drift dataset: PCA
Batch 1, 16 MOS sensors x 8 features = 128 features, 6 gases. Download batch1.dat ... batch10.dat from the UCI page into data/uci/.

```python
import numpy as np, pandas as pd, matplotlib.pyplot as plt
from sklearn.datasets import load_svmlight_file
from sklearn.preprocessing import StandardScaler
from sklearn.decomposition import PCA

X, y = load_svmlight_file('../data/uci/batch1.dat')
X = X.toarray(); print(X.shape, np.bincount(y.astype(int)))
```

```python
Xs = StandardScaler().fit_transform(X)
p = PCA(2).fit(Xs); Z = p.transform(Xs)
names = {1:'ethanol',2:'ethylene',3:'ammonia',4:'acetaldehyde',5:'acetone',6:'toluene'}
for g in np.unique(y):
    plt.scatter(*Z[y==g].T, s=8, label=names[int(g)])
plt.xlabel(f'PC1 ({p.explained_variance_ratio_[0]:.0%})'); plt.ylabel(f'PC2 ({p.explained_variance_ratio_[1]:.0%})')
plt.legend(); plt.savefig('figures/01_pca_batch1.png', dpi=200)
```
