# 02 Drift: train on batch 1, test on later batches

```python
import numpy as np
from sklearn.datasets import load_svmlight_file
from sklearn.pipeline import make_pipeline
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import LogisticRegression

load = lambda b: [a.toarray() if hasattr(a,'toarray') else a for a in load_svmlight_file(f'../data/uci/batch{b}.dat', n_features=128)]
X1, y1 = load(1)
clf = make_pipeline(StandardScaler(), LogisticRegression(max_iter=3000)).fit(X1, y1)
for b in range(2, 11):
    Xb, yb = load(b); print(f'batch {b}: accuracy {clf.score(Xb, yb):.2f}')
```
