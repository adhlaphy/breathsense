```python
from breathlib import *
df = load("data/2026-12-21_s01_breath.csv")
rows = [metrics(w) for w in breaths(df)]
res = pd.DataFrame(rows); print(res.round(3))
res.to_csv("analysis/e3_first_breaths.csv", index=False)

import matplotlib.pyplot as plt
fig, ax1 = plt.subplots(figsize=(8, 3.5))
ax1.plot(df.t, df.gas_ohm / 1000, label="gas resistance (kohm)")
ax2 = ax1.twinx(); ax2.plot(df.t, df.RH_sht, color="gray", alpha=.6, label="RH (%)")
ax1.set_xlabel("time (s)"); ax1.set_ylabel("R (kohm)"); ax2.set_ylabel("RH (%)")
fig.tight_layout(); fig.savefig("analysis/figures/e3_overlay.png", dpi=200)
```
